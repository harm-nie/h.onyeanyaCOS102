```python
# python code to determine annual tax revenue (atr)

# input from the user
years_experience = int(input("enter years of work of experience: "))
age = int(input("enter age: "))

# to determine the atr
if years_experience > 25 and age >= 55:
  atr = 5600000
elif years_experience > 20 and age >= 45:
     atr = 4480000
elif years_experience > 10 and age >= 35:
     atr =1500000
else:
    atr = 550000

# output result
print("annual tax revenue (atr) is: N", atr)
```
